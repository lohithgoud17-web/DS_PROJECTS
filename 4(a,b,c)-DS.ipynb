```python
import pandas as pd
import numpy as np

data = pd.Series(
    np.random.randn(6),
    index=[
        ['a', 'a', 'b', 'b', 'c', 'c'],
        [1, 2, 1, 2, 1, 2]
    ]
)
print(data)

print(data['b'])

print(data[:, 1])
```

```python
import numpy as np
import pandas as pd
data = {
    'A': [1, 2, np.nan],
    'B': [4, np.nan, np.nan]
}
df = pd.DataFrame(data)
print(df.isna())
df_clean = df.dropna()
df_filled = df.fillna(0)
print("\nDataFrame after dropna:")
print(df_clean)
print("\nDataFrame after fillna:")
print(df_filled)
```
