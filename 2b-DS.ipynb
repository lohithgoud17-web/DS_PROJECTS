```python
# SQL Database Operations using SQLite

import sqlite3
import pandas as pd
```

```python
# Connect to database
conn = sqlite3.connect("students.db")
cursor = conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL
)
""")
```

```python
# Insert records
cursor.execute("INSERT INTO students VALUES (1, 'Vinay')")
cursor.execute("INSERT INTO students VALUES (2, 'Panduu')")

conn.commit()
```

```python
# Read data
df = pd.read_sql_query("SELECT * FROM students", conn)
print("Student Records:")
print(df)
```

```python
# Update record
cursor.execute("""
UPDATE students
SET name = 'Sonu'
WHERE id = 1
""")

conn.commit()
```

```python
print("\nAfter Update:")
df = pd.read_sql_query("SELECT * FROM students", conn)
print(df)
```

```python
# Delete record
cursor.execute("""
DELETE FROM students
WHERE id = 1
""")

conn.commit()
```

```python
print("\nAfter Delete:")
df = pd.read_sql_query("SELECT * FROM students", conn)
print(df)
```

```python
# Close connection
conn.close()
```
