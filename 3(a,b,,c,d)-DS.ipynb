```python
import numpy as np
import pandas as pd

data = {'a': [1, np.nan, 3, 4],
        'b': ['one', np.nan, np.nan, 'four']}

df = pd.DataFrame(data)

print(df.isna())

df_clean = df.dropna()

df_filled = df.fillna(0)

print(df_clean)
print(df_filled)
```
